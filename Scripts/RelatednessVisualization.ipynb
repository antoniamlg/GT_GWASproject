# Plot for Visualizing Relatedness

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load relatedness file
genome = pd.read_csv("../data/merged/related/merged_afterqc.genome", delim_whitespace=True)

# Filter to only show PI_HAT values between 0.1 and 0.3
filtered = genome[(genome['PI_HAT'] >= 0.1) & (genome['PI_HAT'] <= 0.3)]

# Plot histogram
plt.figure(figsize=(8, 5))
plt.hist(filtered['PI_HAT'], bins=40, color='#7f63b8', edgecolor='#480c7f')
plt.xlabel('π̂ (PI_HAT)')
plt.ylabel('Number of pairs')
plt.title('Pairwise relatedness (π̂) between 0.1 and 0.3')
plt.axvline(0.185, color='#ba543d', linestyle='--', label='Threshold for relatedness')
plt.legend()
plt.tight_layout()
plt.show()
```
