# Split IIDs by chip

*Input*: metadata.txt <br> *Output*: 5 files named chipX.csv which contain the IDs of all individuals part of one chip

## Import libs

```python
import pandas as pd
from pathlib import Path
import os
```

```python
os.getcwd()
```

```python
# load the metadatafile
metadata = pd.read_csv("../Data/metadata.txt", sep="\t")

# we want IID not user
metadata = metadata.rename(columns={"user": "IID"})
# we don't have FID but plink needs it so ...
metadata["FID"] = metadata["IID"]
# Replace missing chip names with placeholder
metadata["chip"] = metadata["chip"].fillna("unknown_chip")

# create a dir to store the output files
Path("../Data/splitbychip").mkdir(exist_ok=True)

# create one .keep file per chip, named after the chip
for chip_name in metadata["chip"].unique():
    chip_df = metadata[metadata["chip"] == chip_name][["FID", "IID"]]
    # Some chips have spaces in their name -> for safe filenames
    safe_chip_name =  str(chip_name).replace(" ", "_")
    # Write .keep file
    chip_df.to_csv(f"../Data/splitbychip/{safe_chip_name}.keep", sep="\t", index=False, header=False) #Plink does not expect headers
```
