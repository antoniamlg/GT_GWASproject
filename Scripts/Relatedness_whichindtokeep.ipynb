This script works with the .genome file I obtained in the "Relatedness and Duplicates" step and aims to create a text file listing individuals that you want to remove from the dataset due to being too closely related to others. You need to execute 
```bash
plink --bfile merged_afterqc --extract related/merged_afterqc.prune.in --genome --min 0.185 --out merged_afterqc
```
beforehand :)

The following code was obtained from the exercises and is not my work.

````python
# works because IID = FID
# if that is not the case for you -> adapt script :)

import pandas as pd

# Read the .genome file
ibd = pd.read_csv('../data/merged/related/ibdcalc_merged.genome', delim_whitespace=True)

# Get unique members from FID1
members = ibd['FID1'].unique()

# Create DataFrame with two identical columns
df_out = pd.DataFrame({'FID': members, 'IID': members})

# Write to file without headers or row indices
df_out.to_csv('wrong_ibd.txt', sep=' ', index=False, header=False)
````
